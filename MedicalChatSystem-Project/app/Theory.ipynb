RAG stands for Retrieval-Augmented Generation.
It’s an AI architecture that combines retrieval-based methods with generative models (like GPT) to improve accuracy and grounding of responses.

🔹 Small Architecture Overview of RAG

User Query → Input question or request.

Retriever → Searches a knowledge base (vector database, documents, or external API) for the most relevant passages.

Uses embeddings + similarity search (e.g., FAISS, Pinecone, Weaviate).

Augmentation → Retrieved documents are appended to the user query.

Generator (LLM) → A large language model (like GPT, LLaMA, Falcon) generates a response using both:

The user’s question

The retrieved supporting knowledge

Final Answer → Response that is grounded in external data and not just the LLM’s memory.

User Query
    |
    v
[Retriever] ---> Knowledge Base (documents, vector store)
    |
    v
Augmented Query (user query + retrieved context)
    |
    v
[Generator: LLM]
    |
    v
Final Answer

In short: RAG = LLM + Search + Context Injection
